```python
import os
import geopandas as gpd
import pandas as pd

# Fonction pour récupérer tous les shapefiles d'un dossier et ses sous-dossiers
def recuperer_shapefiles(dossier):
    shapefiles = []
    for root, dirs, files in os.walk(dossier):
        for file in files:
            if file.endswith(".shp") and file.startswith("CF_Poly"):
                shapefiles.append(os.path.join(root, file))
    print(len(shapefiles))
    return shapefiles

# Fonction pour fusionner tous les shapefiles dans un seul GeoDataFrame
def fusionner_shapefiles(shapefiles):
    # Charger le premier shapefile
    gdf_total = gpd.read_file(shapefiles[0]).to_crs('WGS 84 / UTM zone 30N')

    # Fusionner les autres shapefiles
    for shapefile in shapefiles[1:]:
        gdf = gpd.read_file(shapefile).to_crs('WGS 84 / UTM zone 30N')
        gdf_total = pd.concat([gdf_total, gdf])

    return gdf_total

# Chemin du dossier où se trouvent les shapefiles
dossier_shapefiles = "/home/userpc/PRESFOR/REGION_TONKPI/CF_TONKPI"

# Récupérer tous les shapefiles
shapefiles = recuperer_shapefiles(dossier_shapefiles)

# Fusionner les shapefiles
gdf_fusionne = fusionner_shapefiles(shapefiles)

# Sauvegarder le résultat fusionné dans un nouveau shapefile
gdf_fusionne.to_file("/home/userpc/PRESFOR/REGION_TONKPI/CF_TONKPI/fichier_fusionne.geojson")
```

```python
import geopandas as gpd
from shapely.ops import unary_union

shapes = gpd.read_file(r"C:\Users\L14\Desktop\PRESFOR\shapefiles_tmci55\villages_delimites_tmci55.shp")

# Réinitialiser l'index pour pouvoir comparer les géométries entre elles
gdf = shapes.reset_index(drop=True)
intersections = []


for i in range(len(gdf)):
    for j in range(i + 1, len(gdf)):
        if gdf.geometry[i].intersects(gdf.geometry[j]):
            inter = gdf.geometry[i].intersection(gdf.geometry[j])
            if not inter.is_empty and inter.geom_type in ['Polygon', 'MultiPolygon']:
                intersections.append(inter)

overlap_gdf = gpd.GeoDataFrame(geometry=intersections, crs=gdf.crs)
overlap_gdf = overlap_gdf[overlap_gdf.geometry.type.isin(['Polygon', 'MultiPolygon'])]

overlap_gdf.to_file(r"C:\Users\L14\Desktop\PRESFOR\shapefiles_tmci55\chevauchements.shp")
```
